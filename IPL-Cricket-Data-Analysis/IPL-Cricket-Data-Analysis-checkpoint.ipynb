# 🏗️ Project Title: IPL Cricket Data Analysis (2008–2023)

 ## Step 1: Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## Step 2: Load the Data

```python
matches = pd.read_csv('matches.csv')
deliveries = pd.read_csv('deliveries.csv')

print(matches.shape)
print(deliveries.shape)
```

## Step 3: Data Cleaning

* Handle null values
* Remove duplicates
* Convert dates
* Standardize team names

```python
matches.isnull().sum()
matches['team1'] = matches['team1'].replace({'Delhi Daredevils':'Delhi Capitals', 'Deccan Chargers':'Sunrisers Hyderabad'})
```

## Step 4: Exploratory Data Analysis (EDA)
### 4.1 Most Successful Team

```python
winner_counts = matches['winner'].value_counts().head(10)
plt.figure(figsize=(8,5))
sns.barplot(x=winner_counts.values, y=winner_counts.index, palette="coolwarm")
plt.title("Most Successful IPL Teams")
plt.xlabel("Wins")
plt.ylabel("Team")
plt.show()
```

### 4.2 Top Players of the Match (MVP)

```python
top_players = matches['player_of_match'].value_counts().head(10)
sns.barplot(x=top_players.values, y=top_players.index)
plt.title("Top 10 Players of the Match Awards")
plt.show()
```

### 4.3 Toss Decision Analysis

```python
sns.countplot(x='toss_decision', data=matches)
plt.title("Toss Decisions (Bat or Field)")
plt.show()
```

### 4.4 Toss vs Win Relationship

```python
toss_win = matches[matches['toss_winner'] == matches['winner']]
percentage = len(toss_win)/len(matches)*100
print(f"Toss winner also won the match: {percentage:.2f}%")
```

### 4.5 Matches Per Season

```python
matches_per_season = matches['season'].value_counts().sort_index()
sns.lineplot(x=matches_per_season.index, y=matches_per_season.values, marker='o')
plt.title("Number of Matches per Season")
plt.xlabel("Season")
plt.ylabel("Matches")
plt.show()
```

### 4.6 Venue Analysis

```python
top_venues = matches['venue'].value_counts().head(10)
sns.barplot(x=top_venues.values, y=top_venues.index)
plt.title("Top Venues by Number of Matches")
plt.show()
```

## Step 5: Player Performance Analysis (Using deliveries.csv)
### 5.1 Top Run Scorers

```python
top_batsmen = deliveries.groupby('batsman')['batsman_runs'].sum().sort_values(ascending=False).head(10)
sns.barplot(x=top_batsmen.values, y=top_batsmen.index)
plt.title("Top 10 Run Scorers")
plt.show()
```

### 5.2 Top Wicket Takers

```python
top_bowlers = deliveries[deliveries['dismissal_kind'].notnull()].groupby('bowler')['dismissal_kind'].count().sort_values(ascending=False).head(10)
sns.barplot(x=top_bowlers.values, y=top_bowlers.index)
plt.title("Top 10 Wicket Takers")
plt.show()
```

### 5.3 Team Batting Performance

```python
team_runs = deliveries.groupby('batting_team')['batsman_runs'].sum().sort_values(ascending=False)
sns.barplot(x=team_runs.values, y=team_runs.index)
plt.title("Total Runs by Teams")
plt.show()
```
